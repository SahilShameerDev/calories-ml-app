```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
url="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
```

```python
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
```

```python
iris_data = pd.read_csv(url, header=None, names=column_names)
```

```python
iris_data.head()
```

```python
iris_data.isnull().sum()
```

```python
iris_data['species'] = iris_data['species'].map({'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2})
```

```python
#Splitting data set

x = iris_data.drop(['species'], axis=1).values
```

```python
y = iris_data['species'].values
```

```python
x
```

```python
y
```

```python
# Standardize the feature values
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=42)
```

```python
#Creating ANN model | Step 1
model = models.Sequential([
    layers.Dense(10, activation='relu', input_shape=(4,)), #input layer
    layers.Dense(4, activation='relu'), #hidden layer
    layers.Dense(3, activation='softmax') #output layer
])

```

```python
model.summary()
```

```python
# Compile the model | Step 2
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Train the model | Step 3
model.fit(X_train, y_train, epochs=45, batch_size=5)
```

```python
#model evaluation
y_pred = np.argmax(model.predict(X_test), axis=-1)
```

```python
y_pred[0]
```

```python
#final step
accuracy = accuracy_score(y_test, y_pred)
```

```python
print(f"Accuracy:{accuracy:.4f}" )
```
